## Point-Charge Model for the Electric Field Gradient (EFG)

As an alternative to density functional theory (DFT), the electric field gradient (EFG) can be approximated using a classical point-charge model, where each ion in the crystal is treated as a point charge located at its crystallographic position.

The electrostatic potential at a probe position $\mathbf{r}_0$ is

$$ V(\mathbf{r}_0)=\frac{1}{4\pi\varepsilon_0} \int \frac{\rho(\mathbf{r}')}{|\mathbf{r}'-\mathbf{r}_0|}\,d\tau', $$

where $\rho(\mathbf{r})$ is the charge density. Assuming a collection of point charges,

$$ \rho(\mathbf{r})=\sum_k q_k\delta(\mathbf{r}-\mathbf{r}_k), $$

the potential becomes

$$ V(\mathbf{r}_0)= \frac{1}{4\pi\varepsilon_0} \sum_k \frac{q_k}{R_k}, $$

where

$$ \mathbf{R}_k = \mathbf{r}_0 - \mathbf{r}_k, \qquad R_k = |\mathbf{R}_k|. $$

The EFG tensor is the Hessian of the electrostatic potential,

$$ V_{ij} = \frac{\partial^2 V}{\partial x_i\partial x_j}, $$

which can be written as the sum of the contributions from all point charges,

$$ V_{ij} = \frac{1}{4\pi\varepsilon_0} \sum_k q_k \left( \frac{3R_{k,i}R_{k,j}-\delta_{ij}R_k^2} {R_k^5} \right), $$

where $\delta_{ij}$ is the Kronecker delta. This expression is exactly the one implemented in `point_charge_EFG()`.

### Sternheimer antishielding correction

To approximately account for the polarization of the electronic cloud surrounding the nucleus, the lattice EFG can be corrected using the Sternheimer antishielding factor,

$$ V_{ij}^{\mathrm{total}} = (1-\gamma_\infty)\, V_{ij}^{\mathrm{lattice}}, $$

where $\gamma_\infty$ is the Sternheimer antishielding factor for the nucleus under consideration.

### Quadrupolar interaction

For nuclei with spin

$$ I > \frac{1}{2}, $$

the electric quadrupole interaction contributes an additional term to the Hamiltonian,

$$ \hat{\mathcal{H}}_Q = \sum_{i}^{N_{\mathrm{nuc}}}\frac{eQ^i(1-\gamma_\infty^i)} {\hbar\,2I(2I-1)} \sum_{\alpha\beta} V_{\alpha\beta}^{i} \hat{I}_\alpha^i \hat{I}_\beta^i, $$

where:
* $Q^i$ is the $i$-th nuclear electric quadrupole moment.
* $V_{\alpha\beta}^{i}$ is the external electric field gradient (EFG) at the site of the $i$-th quadrupolar nucleus, and
* $\hat{I}_\alpha^i$ is the $i$-th nuclear spin operators.

Diagonalization of the EFG tensor yields its principal components

$$ (V_{xx},V_{yy},V_{zz}), $$

ordered according to

$$ |V_{zz}|\ge |V_{yy}|\ge |V_{xx}|, $$

from which the asymmetry parameter

$$ \eta=\frac{V_{xx}-V_{yy}}{V_{zz}}, $$

and the quadrupole coupling constant

$$ \chi_Q = \frac{eQV_{zz}}{h} $$

are readily obtained.
